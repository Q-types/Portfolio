```python
# Idetify mixed columns
cols_mixed = ['Total Revenue',
                   'Unit Cost',
                   'Discount',
                   'Profit']
```

```python
# Define and check each mixed column for brackets enclosing value, then remove
for col in cols_mixed:
  # Define a mask to filter entries where profit contains brackets
  mask_b = df[col].astype(str).str.match(r'^\(.*\)$', na=False)
  mask_b1 = df[col].astype(str).str.match(r'^\(', na=False)

  # Mask to filter where columns contains $
  mask_d = df[col].astype(str).str.match(r'^\$', na=False)
  mask_d1 = df[col].astype(str).str.contains(r'^\$')
  
  print(f'For column: {col} \n - all brackets enclose values: {mask_b.equals(mask_b1)}')
  print(f'- all dollar signs at start: {mask_d.equals(mask_d1)}')

col = 'Discount'

# Mask for % in discount
mask_p = df[col].astype(str).str.endswith('%', na=False)
mask_p1 = df[col].astype(str).str.contains(r'%', na=False)

# Mask for Discount containing .
mask_dec = df[col].astype(str).str.contains(r'\.', na=False)
mask_dec1 = df[col].astype(str).str.strip().eq('.')

print(f'For column: {col} \n - all percentages at end: {mask_p.equals(mask_p1)}')
print(f'- all decimals present are entire entries: {mask_dec.equals(mask_dec1)}')

```

This shows it is safe to replace all values without missing some entries with alternative/mistakes using the defined punctuation.

```python
# Check for na's before transfromation
df.isna().sum()
```

```python
# Replacing punctuation

for col in cols_mixed:
  s = df[col].astype(str).str.strip()
  # Remove currency.percentage anywhere
  s = s.str.replace(r'\$', '', regex=True)
  s = s.str.replace(r'%', '', regex=True)
  # Remove thousands separators
  s = s.str.replace(r',', '', regex=False)
  # Replace parentheses with the enclosed value
  s = s.str.replace(r'^\((.*?)\)', r'\1', regex=True)
  # Normalise lone placeholders
  s = s.mask(s.str.fullmatch(r'\.'), '0')
  # Set dataframe column to the transformed column
  df[col] = pd.to_numeric(s, errors='coerce')
  
```

## Feature Engineering

**Features**
- frequency
- recency

**Restrict Analysis to the 5 newly generated features**

Restricting analysis initially to the engineered features.

```python
# Define end date
end_date = df['Order_Date'].max()

# Aggregate data so each row is one customer
# Create features 
df_customer = (
    df.groupby('Customer ID').agg(
          Frequency=('Order ID', 'count'),
          Recency=('Order_Date', lambda x: (end_date - x.max()).days),
          CLV=('Profit', 'sum'),
          AUC=('Unit Cost', 'mean'),
          Customer_Age=('Customer_BirthDate',
                        lambda x: (end_date - pd.to_datetime(x.iloc[0])).days // 365)
      )
      .reset_index()
)
```

## Feature Scaling and Encoding

**Frequency, CLV, AUC, Recency**

A $\log(1 + x)$ transformation will be applied to these features to compress long right-tailed distributions, stabilise variance, and reduce the disproportionate influence of extreme values when computing Euclidean distances in k-means clustering.

**Age**

No transformation will be applied, as the distribution is approximately symmetric and bounded. The feature is scaled only to ensure comparability with other inputs during distance calculations.

```python
# Import libraries
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer, StandardScaler
```

```python
# List the columns for each method
cols_log = ['Frequency', 'CLV', 'AUC', 'Recency']
col_age = ['Customer_Age']
```

```python
# Define piplines
pipe_log = Pipeline(steps=[
  ('log1p', FunctionTransformer(np.log1p, feature_names_out='one-to-one', validate=False)),
  ('scale', StandardScaler())
])

pipe_age = Pipeline(steps=[
  ('scale', StandardScaler())
])

preprocessor = ColumnTransformer(
  transformers=[
    ('log', pipe_log, cols_log),
    ('other', pipe_age, col_age)
  ]
)
```

## KMeans Clustering

**Optimising the number of clusters**

We now determine the optimum number of clusters, k. This is evaluated using three complementary methods:
- Elbow method
- Silhouette score
- Hierarchical clustering

By assessing the results of each approach and identifying areas of agreement between them, a consensus can be reached on a suitable value of k that balances model performance, interpretability, and business actionability.

**Elbow Method**

```python
# Import libraries
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Set a range for cluster number
Ks = range(2,20)

# Create a list to track inertia
inertias = []

# Loop through valeuas of k
for k in Ks:
  pipe = Pipeline(steps=[
    ('prep', preprocessor),
    ('kmeans', KMeans(n_clusters=k, random_state=42, n_init='auto'))
  ])
  pipe.fit(df_customer)
  inertias.append(pipe['kmeans'].inertia_)

# Plot the results
plt.plot(Ks, inertias, marker='o')
plt.xticks(Ks)
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()
```

The elbow plot displays three clear phases:
1. Steep drop from k=2 to k=4
2. Moderate decline, with diminishing returns from k=4 to k=7
3. Near linear tail beyond k=8

The elbow is around k=3 to k=5, suggesting 3-5 clusters as optimal.

**Silhouette Method**

```python
# Import library
from sklearn.metrics import silhouette_score

# Create a list to track scores
scores = []

# Loop through values of k
for k in Ks:
  pipe = Pipeline(steps=[
    ('prep', preprocessor),
    ('kmeans', KMeans(n_clusters=k, random_state=42, n_init='auto'))
  ])
  
  pipe.fit(df_customer)

  X = pipe['prep'].transform(df_customer)
  labels = pipe['kmeans'].labels_

  score = silhouette_score(X, labels)
  scores.append(score)

# Plot the results
plt.plot(Ks, scores, marker='o')
plt.xticks(Ks)
plt.xlabel("Number of clusters (k)")
plt.ylabel("Silhouette score")
plt.title("Silhouette Method")
plt.show()
```

The silhouette method indicated that the highest average separation occurred for k=2, with a comparable score observed at k=3. Beyond k=3, the silhouette score declined steadily, suggesting diminishing returns from additional clusters. Although k=2 provides the strongest geometric separation, a three-cluster solution was selected as it balances quantitative cluster quality with greater interpretability and actionability for marketing segmentation.

**Hierarchical Clustering**

The data naturally supports one major split, and then a meaningful secondary subdivision, yielding three coherent groups before fragmentation begins.

**Optimum Clusters K**

The Elbow method indicated a clear reduction in marginal inertia improvement beyond k = 4, with the most pronounced change in curvature occurring between k = 3 and k = 4. Consistent with this, the silhouette analysis achieved its highest scores for k = 2 and k = 3, followed by a steady decline as additional clusters were introduced, suggesting diminishing returns from further segmentation.

Hierarchical clustering using Ward linkage was performed to complement the k-means analysis. The resulting dendrogram revealed a small number of dominant branches with clear separation at higher linkage distances, followed by meaningful secondary subdivisions prior to finer fragmentation. This structure supports a choice of 2-3 clusters.

Based on the agreement between the Elbow, Silhouette, and hierarchical clustering methods, and considering the need for actionable customer segmentation, a three-cluster solution (k = 3) was selected. This choice captures the dominant structure in the data while maintaining meaningful separation between customer groups and avoiding over-segmentation.

**Cluster 1**

**Cluster 2**
